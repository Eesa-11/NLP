Task 04: Sets for Unique Vocabulary
Topic: Set (set) Core Python Data Structure
(a)  Create a list called raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]. Convert it into a set called vocab. Print both raw and vocab, and print the number of unique words. Explain in a comment why sets remove duplicates automatically.
(b)  Create set_A = {"machine", "learning", "data", "model"} and set_B = {"deep", "learning", "neural", "data"}. Compute and print: (i) the union of both sets, (ii) the intersection (words in common), and (iii) the difference set_A − set_B (words only in A).
(c)  Define a stopwords set: stopwords = {"is", "the", "a", "in", "of", "and", "to"}. Given the token list tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"], use set difference to remove all stopwords and print the cleaned tokens.
(d)  Starting from your vocab set in part (a), add the words "transformer" and "bert" using the appropriate method. Remove "python" using the discard() method. Then use the in operator to check whether "ai" is still present, and print the final set.

```python
#TASK 04

# A)
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab = set(raw)
print("Raw list:", raw)
print("Vocabulary set:", vocab)
print("Number of unique words:", len(vocab))
# Sets are unordered collections of unique elements. When a list is converted to a set,
# duplicate elements are automatically removed because a set cannot contain identical items.
# This property makes sets ideal for efficiently finding unique elements in a collection.

# B)
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

# Union
union_set = set_A.union(set_B)
print("\nUnion of set_A and set_B:", union_set)

# Intersection
intersection_set = set_A.intersection(set_B)
print("Intersection of set_A and set_B:", intersection_set)

# Difference (words only in A)
difference_set = set_A.difference(set_B)
print("Difference (set_A - set_B):", difference_set)

# C)
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

# Convert tokens to a set to perform set difference, then convert back to list for cleaned_tokens
cleaned_tokens = list(set(tokens) - stopwords)
print("\nCleaned tokens (stopwords removed):", cleaned_tokens)

# D)
vocab.add("transformer")
vocab.add("bert")
vocab.discard("python")

print("\nIs 'ai' present in vocab?", "ai" in vocab)
print("Final vocabulary set:", vocab)
```

(d)  Create a tuple called pipeline_steps = ("load_data", "clean_text", "tokenize", "remove_stopwords", "vectorize"). Print the full pipeline. Use the index() method to find the position of "tokenize". Print the total number of steps, and demonstrate that the tuple cannot be modified by trying to delete one element inside a try-except block.

```python
#TASK 05

# A)
pos_tags = [("language", "NOUN"), ("processes", "VERB"), ("human", "ADJ"), ("text", "NOUN"), ("quickly", "ADV")]
print("Full list of POS tags:", pos_tags)

# Access second tuple (index 1)
second_tuple = pos_tags[1]
print("Word from second tuple:", second_tuple[0])
print("Tag from second tuple:", second_tuple[1])

# B)
# Unpack the first tuple
word, tag = pos_tags[0]
print("\nUnpacked word from first tuple:", word)
print("Unpacked tag from first tuple:", tag)

# Attempt to change the tag of the first tuple
try:
    pos_tags[0][1] = "VERB"
except TypeError as e:
    print("Error attempting to modify tuple:", e)

# C)
def extract_nouns(pos_list):
    nouns = []
    for word, tag in pos_list:
        if tag == "NOUN":
            nouns.append(word)
    return nouns

nouns_list = extract_nouns(pos_tags)
print("\nExtracted nouns:", nouns_list)

# D)
pipeline_steps = ("load_data", "clean_text", "tokenize", "remove_stopwords", "vectorize")
print("\nFull pipeline:", pipeline_steps)

# Find position of 'tokenize'
tokenize_position = pipeline_steps.index("tokenize")
print("Position of 'tokenize':", tokenize_position)

# Total number of steps
print("Total number of steps:", len(pipeline_steps))

# Demonstrate immutability by trying to delete an element
try:
    del pipeline_steps[0] # Tuples do not support item deletion
except TypeError as e:
    print("Error attempting to delete element from tuple:", e)
```

Task 06: String Cleaning Methods
Topic: String Methods lower, upper, strip, split, join, replace
(a)  Given text = "   MACHINE Learning is AMAZING!!!   ", apply these methods one by one and print the result after each step: (i) strip() to remove leading and trailing spaces, (ii) lower() to normalise case, (iii) replace() to remove all "!" characters.
(b)  Take the cleaned text from part (a). Use split() to tokenise it into a list of words and print the list. Then use upper() on each word inside a list comprehension and print the uppercase list. Finally, use join() with " | " as separator to combine the uppercase words back into a single string.
(c)  Create the string sentence = "I love Java and Java is popular". Use replace() to substitute every occurrence of "Java" with "Python". Then use find() to locate the starting position of the first occurrence of "Python" in the updated sentence and print it.
(d)  Write a function called normalize(text) that takes a raw string and returns a clean version by: stripping whitespace, converting to lowercase, replacing commas and full-stops with an empty string. Test the function on at least two messy input strings and print the cleaned output for each.

```python
#TASK 06

# A)
text = "   MACHINE Learning is AMAZING!!!   "
print("Original text:", text)

# (i) strip()
stripped_text = text.strip()
print("After strip():", stripped_text)

# (ii) lower()
lowercase_text = stripped_text.lower()
print("After lower():", lowercase_text)

# (iii) replace()
cleaned_text_a = lowercase_text.replace("!", "")
print("After replace('!', ''):", cleaned_text_a)

# B)
# Use the cleaned text from part (a)
print("\nCleaned text from (a) for part (b):", cleaned_text_a)

# (i) split()
tokens_b = cleaned_text_a.split()
print("After split():", tokens_b)

# (ii) upper() on each word with list comprehension
uppercase_tokens = [word.upper() for word in tokens_b]
print("Uppercase words:", uppercase_tokens)

# (iii) join()
joined_string = " | ".join(uppercase_tokens)
print("Joined string:", joined_string)

# C)
sentence_c = "I love Java and Java is popular"
print("\nOriginal sentence (c):", sentence_c)

# replace()
updated_sentence_c = sentence_c.replace("Java", "Python")
print("After replace():", updated_sentence_c)

# find()
python_position = updated_sentence_c.find("Python")
print("Position of first 'Python':", python_position)

# D)
def normalize(text):
    # Strip whitespace, convert to lowercase
    clean_text = text.strip().lower()
    # Replace commas and full-stops with empty string
    clean_text = clean_text.replace(",", "").replace(".", "")
    return clean_text

print("\nTesting normalize function:")
# Test case 1
input1 = "  Hello, World! This is a TEST.  "
output1 = normalize(input1)
print(f"Input: '{input1}'\nOutput: '{output1}'")

# Test case 2
input2 = "NLP is AMAZING, but sometimes messy."
output2 = normalize(input2)
print(f"Input: '{input2}'\nOutput: '{output2}'")
```

Task 07: String Validation Methods
Topic: String Methods find, startswith, endswith, isdigit, isalpha
(a)  Given text = "NLP is the future of AI research", use startswith() to check whether the text begins with "NLP", and endswith() to check whether it ends with "research". Print a clear message for each result. Then use find() to get the position of the word "future" in the string.
(b)  Create a mixed list: tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]. Loop through the list and for each token print whether it is purely alphabetic (isalpha()), purely numeric (isdigit()), or mixed. Label each output clearly.
(c)  Given the list words = ["running", "jumped", "NLP", "ai2025", "data", "42"]. Use a loop with isalpha() and isdigit() to separate the list into three groups: pure_words, pure_numbers, and mixed_tokens. Print each group.
(d)  Write a function called validate_token(token) that returns the string "WORD" if the token is purely alphabetic, "NUMBER" if it is purely numeric, and "MIXED" otherwise. Apply this function to each element in the list ["chat", "99", "gpt4", "bert", "2024", "nlp!"] using a loop and print the token alongside its label.

```python
#TASK 07

# A)
text_a = "NLP is the future of AI research"

# Check startswith()
if text_a.startswith("NLP"):
    print(f"The text '{text_a}' starts with 'NLP'.")
else:
    print(f"The text '{text_a}' does not start with 'NLP'.")

# Check endswith()
if text_a.endswith("research"):
    print(f"The text '{text_a}' ends with 'research'.")
else:
    print(f"The text '{text_a}' does not end with 'research'.")

# Find position of 'future'
future_position = text_a.find("future")
print(f"The word 'future' is found at index: {future_position}")

# B)
tokens_b = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
print("\n--- Token Analysis (Part B) ---")
for token in tokens_b:
    if token.isalpha():
        print(f"'{token}' is purely alphabetic.")
    elif token.isdigit():
        print(f"'{token}' is purely numeric.")
    else:
        print(f"'{token}' is mixed (alphanumeric or special characters).")

# C)
words_c = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words_c:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("\n--- Token Grouping (Part C) ---")
print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)

# D)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

tokens_d = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
print("\n--- Token Validation (Part D) ---")
for token in tokens_d:
    label = validate_token(token)
    print(f"Token: '{token}' -> Type: {label}")
```

Task 08: Functions and Lambda for NLP Pipelines
Topic: Functions, Lambda Reusable Text Processing
(a)  Write a regular function called clean(text) that takes a string, strips whitespace, converts it to lowercase, and returns the result. Then write an equivalent lambda function called clean_lambda that performs the same two operations in one line. Test both on the string "  HELLO Natural Language Processing  " and print their outputs.
(b)  Write a function called tokenize(text) that takes a cleaned string and returns a list of individual word tokens using split(). Then write another function called remove_short(tokens, n) that takes the token list and an integer n, and returns only tokens whose length is greater than n. Chain both functions on the sentence "I love working with NLP tools in Python daily" with n = 3 and print the result.
(c)  Write a function called word_lengths(tokens) that takes a list of tokens and returns a dictionary where each key is a word and each value is the length of that word. Call the function on tokens = ["language", "model", "ai", "processing", "deep"] and print the resulting dictionary.
(d)  Combine the functions from parts (b) and (c) into a full pipeline function called nlp_pipeline(text, min_len) that: cleans the text, tokenises it, removes short words, and returns a length dictionary. Call this pipeline on two different sentences and print the outputs.

```python
#TASK 08

# A)
# Regular function for cleaning text
def clean(text):
    return text.strip().lower()

# Lambda function for cleaning text
clean_lambda = lambda text: text.strip().lower()

test_string = "  HELLO Natural Language Processing  "
print(f"Original string: '{test_string}'")
print(f"Cleaned (regular function): '{clean(test_string)}'")
print(f"Cleaned (lambda function): '{clean_lambda(test_string)}'")

# B)
# Function to tokenize a string
def tokenize(text):
    return text.split()

# Function to remove short tokens
def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]

sentence_b = "I love working with NLP tools in Python daily"
n_b = 3

# Chain the functions
cleaned_sentence_b = clean(sentence_b) # Reusing clean from part (a)
tokens_b = tokenize(cleaned_sentence_b)
filtered_tokens_b = remove_short(tokens_b, n_b)
print(f"\nOriginal sentence (b): '{sentence_b}'")
print(f"Tokens longer than {n_b} characters: {filtered_tokens_b}")

# C)
# Function to get word lengths
def word_lengths(tokens):
    return {token: len(token) for token in tokens}

tokens_c = ["language", "model", "ai", "processing", "deep"]
word_lengths_dict = word_lengths(tokens_c)
print(f"\nWord lengths dictionary: {word_lengths_dict}")

# D)
# Combined NLP pipeline function
def nlp_pipeline(text, min_len):
    cleaned_text = clean(text) # Use clean from part (a)
    tokens = tokenize(cleaned_text) # Use tokenize from part (b)
    filtered_tokens = remove_short(tokens, min_len) # Use remove_short from part (b)
    length_dict = word_lengths(filtered_tokens) # Use word_lengths from part (c)
    return length_dict

print("\n--- Testing NLP Pipeline (Part D) ---")
# Test case 1
sentence_d1 = "  Natural Language Processing is a fascinating field.  "
min_len_d1 = 2
pipeline_output1 = nlp_pipeline(sentence_d1, min_len_d1)
print(f"Sentence 1: '{sentence_d1}'")
print(f"Pipeline output (min_len={min_len_d1}): {pipeline_output1}")

# Test case 2
sentence_d2 = "Python powers many AI applications."
min_len_d2 = 5
pipeline_output2 = nlp_pipeline(sentence_d2, min_len_d2)
print(f"\nSentence 2: '{sentence_d2}'")
print(f"Pipeline output (min_len={min_len_d2}): {pipeline_output2}")
```

Task 09: map(), filter(), sorted() and Counter
Topic: map(), filter(), sorted(), Lambda, collections.Counter
(a)  Given words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]. Use map() with a lambda to: (i) convert every word to lowercase — print the result as a list, and (ii) compute the character length of each word — print the result as a list.
(b)  Using the same words list, apply filter() with a lambda to create two filtered lists: (i) words_long — only words with more than 3 characters, and (ii) words_with_t — only words containing the letter "t" (case-insensitive). Print both lists.
(c)  Sort the words list using sorted() in three different ways and print each result with a label: (i) alphabetically (default order), (ii) by word length ascending using key=len, (iii) by word length descending using reverse=True.
(d)  Import Counter from collections. Given the sentence "nlp nlp python ai nlp python text ai nlp python text", split it into words and create a Counter called freq. Print the full counter, then print: (i) the 3 most common words using most_common(), (ii) the frequency of the word "python", and (iii) the total number of words and the size of the unique vocabulary.

```python
#TASK 09

# A)
words_a = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (i) Convert to lowercase using map() and lambda
lowercase_words = list(map(lambda word: word.lower(), words_a))
print(f"Lowercase words: {lowercase_words}")

# (ii) Compute character length using map() and lambda
word_lengths_a = list(map(lambda word: len(word), words_a))
print(f"Word lengths: {word_lengths_a}")

# B)
words_b = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (i) Words with more than 3 characters using filter() and lambda
words_long = list(filter(lambda word: len(word) > 3, words_b))
print(f"\nWords longer than 3 characters: {words_long}")

# (ii) Words containing 't' (case-insensitive) using filter() and lambda
words_with_t = list(filter(lambda word: 't' in word.lower(), words_b))
print(f"Words containing 't' (case-insensitive): {words_with_t}")

# C)
words_c = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (i) Alphabetically (default order)
sorted_alpha = sorted(words_c)
print(f"\nSorted alphabetically: {sorted_alpha}")

# (ii) By word length ascending
sorted_by_length_asc = sorted(words_c, key=len)
print(f"Sorted by length (ascending): {sorted_by_length_asc}")

# (iii) By word length descending
sorted_by_length_desc = sorted(words_c, key=len, reverse=True)
print(f"Sorted by length (descending): {sorted_by_length_desc}")

# D)
from collections import Counter

sentence_d = "nlp nlp python ai nlp python text ai nlp python text"
words_d = sentence_d.split()

freq = Counter(words_d)
print(f"\nFull word frequency counter: {freq}")

# (i) 3 most common words
most_common_words = freq.most_common(3)
print(f"3 most common words: {most_common_words}")

# (ii) Frequency of 'python'
python_freq = freq['python']
print(f"Frequency of 'python': {python_freq}")

# (iii) Total number of words and unique vocabulary size
total_words = sum(freq.values())
unique_vocab_size = len(freq)
print(f"Total number of words: {total_words}")
print(f"Size of unique vocabulary: {unique_vocab_size}")
```
