### TASK 01

```python
# TASK 01: Regex Fundamentals

import re

text = "Hello World! I have 2 apples, 25 books, year 2024, code 1234, and $50. Python NLP is useful."

# A) Extract all digits
pattern = r'\d'
print("A:", re.findall(pattern, text))

# B) Extract all words that start with a capital letter
pattern = r'\b[A-Z][a-zA-Z]*\b'
print("B:", re.findall(pattern, text))

# C) Extract all sequences of exactly 4 digits
pattern = r'(?<!\d)\d{4}(?!\d)'
print("C:", re.findall(pattern, text))

# D) Extract all words between 3 to 6 characters
pattern = r'\b[A-Za-z]{3,6}\b'
print("D:", re.findall(pattern, text))
```

### TASK 02

```python
# TASK 02: Pattern-Based Extraction

text = """Contact ali.khan@example.com or support@test.org.
Call 0300-1234567 or 0312-7654321.
Order IDs are ORD-2024-00123 and ORD-2025-00456.
Dates are 15/08/2024 and 22/09/2025.
Sales increased by 45% and discount was 12.5%."""

# A) Email addresses
pattern = r'[\w.-]+@[\w-]+\.[a-zA-Z]{2,}'
print("A:", re.findall(pattern, text))

# B) Phone numbers
pattern = r'03\d{2}-\d{7}'
print("B:", re.findall(pattern, text))

# C) Dates
pattern = r'\d{2}/\d{2}/\d{4}'
print("C:", re.findall(pattern, text))

# D) Percentage values
pattern = r'\d+(?:\.\d+)?%'
print("D:", re.findall(pattern, text))
```

### TASK 03

```python
# TASK 03: Word Tokenization

import nltk
import spacy
from nltk.tokenize import word_tokenize

# Download required NLTK resource
nltk.download('punkt_tab')

paragraph = """I don't like noisy places.
It's easier to study at home.
We can read books and practice NLP there.
Sometimes, we don't have enough time, but we still try."""

# A) NLTK
nltk_tokens = word_tokenize(paragraph)
print("A) NLTK tokens:")
print(nltk_tokens)

# B) spaCy
nlp = spacy.blank("en")
doc = nlp(paragraph)
spacy_tokens = [token.text for token in doc]

print("\nB) spaCy tokens:")
print(spacy_tokens)

# C) Comparison
# NLTK splits contractions such as "don't" into "do" and "n't".
# spaCy keeps "don't" as one token.
# Both methods keep punctuation as separate tokens.

print("\nC) Comparison:")
print("NLTK splits contractions differently from spaCy.")

# D) Token counts
print("\nD) Token counts:")
print("NLTK:", len(nltk_tokens))
print("spaCy:", len(spacy_tokens))
```

### TASK 04

```python
# TASK 04: Token Attributes

nlp = spacy.blank("en")

sentence = "Ali has 50 dollars and his email is ali@example.com. He paid $50."
doc = nlp(sentence)

# A) Load sentence into spaCy Doc
print("A) Doc:")
print(doc)

# B) is_alpha
print("\nB) is_alpha:")
for token in doc:
    print(token.text, token.is_alpha)

# C) like_num and like_email
print("\nC) like_num and like_email:")
for token in doc:
    print(token.text, token.like_num, token.like_email)

# D) Currency tokens
print("\nD) Currency tokens:")
currency_tokens = [token.text for token in doc if token.is_currency]
print(currency_tokens)
```

### TASK 05

```python
# TASK 05: Sentence Segmentation

# A) Load trained spaCy pipeline
nlp = spacy.load("en_core_web_sm")

paragraph = """Dr. Ali went to the market in the morning.
He bought fruits and vegetables.
Mr. Khan joined him later, e.g. after the meeting ended.
They returned home safely."""

doc = nlp(paragraph)

# B) Sentence segmentation
print("B) Sentences:")
sentences = list(doc.sents)

for sent in sentences:
    print(sent.text)

# C) Total number of sentences
print("\nC) Total sentences:", len(sentences))

# D) Observation
# The trained spaCy model does not incorrectly split at "Dr.", "Mr.", or "e.g." here.
print("\nD) Observation:")
print("The abbreviations did not cause incorrect sentence breaks.")
```
